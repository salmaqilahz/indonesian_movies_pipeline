# Explore IMDb staging output

Checks on the raw IMDb extraction (titles with an Indonesian *release region*):
- how many records, and any duplicates
- breakdown by titleType (movie / tvSeries / tvMiniSeries / etc)
- field completeness

Reminder: this is release-region data, not confirmed production origin.
Some titles here may not actually be Indonesian productions — just released there.

```python
import json
from pathlib import Path
from collections import Counter

STAGING_DIR = Path("../staging_data/imdb")
latest_file = sorted(STAGING_DIR.glob("imdb_raw_*.json"))[-1]
print(f"Loading: {latest_file.name}")

with open(latest_file, encoding="utf-8") as f:
    data = json.load(f)

print(f"Total rows: {len(data)}")
```

```python
# Check for duplicate tconst
tconsts = [r["tconst"] for r in data]
unique_tconsts = set(tconsts)
print(f"Total: {len(tconsts)}, unique: {len(unique_tconsts)}, duplicates: {len(tconsts) - len(unique_tconsts)}")
```

```python
# Breakdown by titleType
type_counts = Counter(r.get("titleType", "?") for r in data)
for title_type, count in type_counts.most_common():
    print(f"{title_type:15s} {count}")
```

```python
# Field completeness
total = len(data)
fields = ["startYear", "runtimeMinutes", "genres", "originalTitle"]
for field in fields:
    present = sum(1 for r in data if r.get(field) is not None)
    print(f"{field:20s} {present}/{total}  ({present/total:.1%} present)")
```
